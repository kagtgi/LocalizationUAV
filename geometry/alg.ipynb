Here is the concise, academic-style pseudocode for the algorithm:

**Algorithm: Region Expansion (Active Frontier)**

**Input:** Dual graph $G$, seed triangle index $s$, max iterations $N$
**Output:** Maximal polygon region $R$, boundary attributes, original vertices $V_{orig}$

1: $V_{orig} \leftarrow \text{Vertices}(s)$

2: $R \leftarrow \{s\}$

3: $F \leftarrow \text{Neighbors}(s, G)$ \quad $\triangleright$ \textit{Frontier set}

4: $V \leftarrow F \cup \{s\}$ \quad $\quad \;\; \triangleright$ \textit{Visited set}

5:

6: **for** $i = 1$ **to** $N$ **do**

7: $\quad added\_flag \leftarrow \text{False}$

8:

9: $\quad$ **for** each candidate $c$ **in** $\text{Sort}(F)$ **do**

10: $\quad \quad R_{temp} \leftarrow R \cup \{c\}$

11: $\quad \quad B_{edges}, B_{verts} \leftarrow \text{ComputeBoundary}(R_{temp})$

12:

13: $\quad \quad$ $\triangleright$ \textit{Constraints Check}

14: $\quad \quad is\_critical \leftarrow V_{orig} \subseteq B_{verts}$ \quad $\triangleright$ \textit{Original vertices remain on boundary}

15: $\quad \quad no\_holes \leftarrow \text{IsSimpleLoop}(B_{edges}, B_{verts})$ \quad $\triangleright$ \textit{Topological validity}

16:

17: $\quad \quad$ **if** $is\_critical$ **and** $no\_holes$ **then**

18: $\quad \quad \quad R \leftarrow R_{temp}$

19: $\quad \quad \quad \text{Remove } c \text{ from } F$

20: $\quad \quad \quad \text{Add unvisited neighbors of } c \text{ to } F \text{ and } V$

21: $\quad \quad \quad added\_flag \leftarrow \text{True}$

22: $\quad \quad \quad$ **break** \quad $\triangleright$ \textit{Restart frontier evaluation with updated region}

23:

24: $\quad$ **if not** $added\_flag$ **then**

25: $\quad \quad$ **break** \quad $\triangleright$ \textit{Stopping criterion: Criticality condition met}

26:

27: **return** $R, \text{ComputeBoundary}(R), V_{orig}$
